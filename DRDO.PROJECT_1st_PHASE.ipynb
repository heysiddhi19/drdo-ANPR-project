```python
import numpy as np
import cv2
from ultralytics import YOLO
from paddleocr import PaddleOCR
import imageio
from google.colab import files

# Initialization
model = YOLO('best.pt')
ocr = PaddleOCR(use_angle_cls=False, lang='en', show_log=False)

cap = cv2.VideoCapture('sample3.mp4')
writer = imageio.get_writer('final_locked_accurate.mp4', fps=20)

history_data = {}
final_locks = {}

while cap.isOpened():
    ret, frame = cap.read()
    if not ret: break

    results = model.track(frame, persist=True, conf=0.3, tracker="bytetrack.yaml", verbose=False)

    if results[0].boxes.id is not None:
        boxes = results[0].boxes.xyxy.cpu().numpy().astype(int)
        ids = results[0].boxes.id.cpu().numpy().astype(int)

        for box, track_id in zip(boxes, ids):
            x1, y1, x2, y2 = box

            if track_id not in final_locks:
                plate = frame[max(0, y1-10):min(frame.shape[0], y2+10),
                              max(0, x1-10):min(frame.shape[1], x2+10)]
                plate = cv2.resize(plate, None, fx=2.5, fy=2.5, interpolation=cv2.INTER_CUBIC)

                result = ocr.ocr(plate, cls=False)
                if result and result[0]:
                    for line in result[0]:
                        txt = line[1][0]
                        conf = line[1][1]
                        if len(txt) >= 5 and conf > 0.5:
                            if track_id not in history_data: history_data[track_id] = []
                            history_data[track_id].append({"text": txt, "score": conf})

                if track_id in history_data and len(history_data[track_id]) >= 7:
                    data = history_data[track_id]
                    best_match = max(data, key=lambda x: x['score'])['text']
                    final_locks[track_id] = best_match

            text_to_show = final_locks.get(track_id, "Processing...")
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(frame, f"ID:{track_id} {text_to_show}", (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    writer.append_data(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

writer.close()
cap.release()
files.download('final_locked_accurate.mp4')
```
